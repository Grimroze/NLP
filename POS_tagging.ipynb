```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
doc = nlp("hey there how are you ?")
```

```python
doc.text
```

```python
doc[0]
```

```python
doc[4].pos_
```

```python
doc[3].tag_
```

```python
spacy.explain("VBP")
```

```python
for word in doc:
  print(word.text, word.pos_, word.tag_, spacy.explain(word.tag_))
```
